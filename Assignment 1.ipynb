```python
import networkx as nx
import matplotlib.pyplot as plt
import nltk
```

```python
# Create nodes and edges from file
G = nx.Graph()
G.add_nodes_from(["Andre", "Beverly", "Carol", "Diane", "Ed", "Fernando", "Garth", "Heather", "Ike", "Jane"])
G.add_edges_from([
    ("Andre", "Beverly"),
    ("Andre", "Diane"), 
    ("Andre", "Carol"),
    ("Andre", "Fernando"),
    ("Beverly", "Ed"),
    ("Beverly", "Garth"),
    ("Beverly", "Diane"),
    ("Carol", "Diane"),
    ("Carol", "Fernando"),
    ("Diane", "Garth"),
    ("Diane", "Ed"),
    ("Diane", "Fernando"),
    ("Ed", "Garth"),
    ("Garth", "Fernando"),
    ("Garth", "Heather"),
    ("Fernando", "Heather"),
    ("Heather", "Ike"), 
    ("Ike", "Jane")
])
```

```python
# Draw kite
pos = nx.spring_layout(G, seed=12)

nx.draw(G, pos, with_labels=True)

plt.axis("off")
plt.show()
```

```python
# List nodes and edges 

print("Nodes:", G.nodes())
print("Edges:", G.edges())
```
